````python
# ═══════════════════════════════════════════════════════════════
#  EJERCICIO 5: ECDSA — Verificar a Mano
# ═══════════════════════════════════════════════════════════════
#
#  Dada la ecuación de verificación ECDSA:
#    R' = u1×G + u2×P
#  donde u1 = z/s mod N y u2 = r/s mod N
#
#  Completa la demostración de que R' = R (el punto nonce original).
#  
#  Partiendo de: R' = (z/s)×G + (r/s)×P
#  Sustituye P = d×G:
#    R' = (z/s)×G + (r/s)×(d×G)
#       = _____________ × G        ← factoriza G
#  Como s = k⁻¹(z + r×d):
#    (z + r×d)/s = (z + r×d) / (k⁻¹(z + r×d)) = _____
#  Por lo tanto R' = _____ × G = R    QED

factored = ""       # <-- ¿Cuál es el escalar antes de ×G? (ej., "(z+rd)/s")
simplified = ""     # <-- ¿A qué se simplifica (z+rd)/s?

correct_factored = "(z+rd)/s"
correct_simplified = "k"

if factored.replace(" ", "") == correct_factored:
    print(f"Ejercicio 5a: ✓  R' = {factored} × G")
else:
    print(f"Ejercicio 5a: ✗  R' = {correct_factored} × G")

if simplified.lower().strip() == correct_simplified:
    print(f"Ejercicio 5b: ✓  Se simplifica a {simplified}, entonces R' = k×G = R  ✓")
else:
    print(f"Ejercicio 5b: ✗  (z+rd)/s se simplifica a k (porque s = (z+rd)/k)")
````

````python
# ═══════════════════════════════════════════════════════════════
#  EJERCICIO 6: Implementar Recuperación de Nonce
# ═══════════════════════════════════════════════════════════════
#
#  Dadas dos firmas (r, s1) y (r, s2) para mensajes con
#  hashes z1 y z2 (¡mismo r significa mismo nonce!), recupera
#  el nonce k y luego la clave privada d.
#
#  Fórmulas:
#    k = (z1 - z2) × (s1 - s2)⁻¹ mod N
#    d = r⁻¹ × (s1×k - z1) mod N

# Configuración (no modificar)
ex6_d = secrets.randbelow(SECP_N - 1) + 1
ex6_k = secrets.randbelow(SECP_N - 1) + 1
ex6_R = scalar_mult(ex6_k, G)
ex6_r = ex6_R.x % SECP_N
ex6_k_inv = pow(ex6_k, SECP_N - 2, SECP_N)

ex6_z1 = int.from_bytes(hashlib.sha256(b"message one").digest(), 'big')
ex6_z2 = int.from_bytes(hashlib.sha256(b"message two").digest(), 'big')
ex6_s1 = (ex6_k_inv * (ex6_z1 + ex6_r * ex6_d)) % SECP_N
ex6_s2 = (ex6_k_inv * (ex6_z2 + ex6_r * ex6_d)) % SECP_N

# TU CÓDIGO: recupera k y d
recovered_k = 0  # <-- Reemplaza con la fórmula
recovered_d = 0  # <-- Reemplaza con la fórmula

# Verificar
if recovered_k == ex6_k:
    print(f"Ejercicio 6a: ✓  Nonce k recuperado!")
else:
    print(f"Ejercicio 6a: ✗  k no recuperado")
    print(f"  Pista: k = (z1 - z2) * pow(s1 - s2, SECP_N - 2, SECP_N) % SECP_N")

if recovered_d == ex6_d:
    print(f"Ejercicio 6b: ✓  Clave privada d recuperada! (Por eso la reutilización del nonce es fatal)")
else:
    print(f"Ejercicio 6b: ✗  d no recuperado")
    print(f"  Pista: d = pow(r, SECP_N - 2, SECP_N) * (s1 * k - z1) % SECP_N")
````
